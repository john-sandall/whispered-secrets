````python
from langchain_community.llms import Ollama

llm = Ollama(model="llama3")
````

````python
print(
    llm.invoke(
        "Hey Ollama, what's your #1 favourite thing about Electromagnetic Field Camp? Use lots of emoji!"
    )
)
````

````python
import requests


@memory.cache
def get_json(url):
    response = requests.get(url)
    response.raise_for_status()
    return response.json()
````

````python
all_talks = get_json("https://www.emfcamp.org/schedule/2024.json")
````

````python
[talk["title"] for talk in all_talks if "Sandall" in talk["speaker"]]
````

````python
talk_data = "\n---\n\n".join([f"{talk['title']} - {talk['speaker']}" for talk in all_talks])

print(talk_data[:500])
````

````python
from IPython.display import Markdown

response = llm.invoke(
    f"""
I'm going to give you a list of talks.
Please can you review all the talk abstracts and recommend the top 5 talks I should go to?
I don't want to know what the website is, just five talk recommendations.

Here's the list:

{talk_data}

---

Top five talk recomendations (formatted in Markdown) are:
""",
)

Markdown(response)
````
